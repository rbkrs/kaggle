```python
import pandas as pd

# Step 1: Load both submissions
best_public = pd.read_csv("best-public-25.csv")
submission = pd.read_csv("attempt-final-level2_stacking.csv")

# Step 2: Merge them on the 'id' column
ensemble = best_public.merge(submission, on="id", suffixes=("_1", "_2"))

# Step 3: Blend the predictions (adjust weights as needed)
ensemble["BeatsPerMinute"] = (
    1.05 * ensemble["BeatsPerMinute_1"] +
    -0.05 * ensemble["BeatsPerMinute_2"]
)

# Step 4: Save final blended submission
final = ensemble[["id", "BeatsPerMinute"]]
final.to_csv("submission-blender.csv", index=False)

print("✅ Ensemble submission saved as submission1.csv")
print(f"🎯 BPM range: {final['BeatsPerMinute'].min():.2f} - {final['BeatsPerMinute'].max():.2f}")
```
