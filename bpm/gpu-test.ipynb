```python
def test_xgboost_gpu():
    """Test XGBoost GPU support"""
    print("\n🌳 Testing XGBoost GPU...")
    try:
        import xgboost as xgb
        import numpy as np
        print(f"✅ XGBoost {xgb.__version__}")
        
        # Test GPU training
        X = np.random.random((1000, 10))
        y = np.random.random(1000)
        
        # Try GPU first
        try:
            dtrain = xgb.DMatrix(X, label=y)
            params = {'tree_method': 'gpu_hist', 'device': 'cuda'}
            model = xgb.train(params, dtrain, num_boost_round=10)
            print("   ✅ GPU training successful")
        except Exception as gpu_error:
            print(f"   ⚠️  GPU training failed: {gpu_error}")
            print("   ℹ️  Falling back to CPU")
        return True
    except Exception as e:
        print(f"❌ XGBoost failed: {e}")
        return False
    
test_xgboost_gpu()
```
